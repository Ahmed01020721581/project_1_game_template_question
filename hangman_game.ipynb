## Description
> ##### The Hangman program randomly selects a secret word from a list of secret words.
> ##### The Game, a random word (a fruit name) is picked up from our collection and the player gets limited chances to win the game.
> ##### Whwn a letter in that word is guessed correctly, that letter position in the word is made visable. In this way, all letters of the word are to be guessed before all the chances are over.
> ##### For convenience, we have given length of word +2 chances. For example, word to be guessed is mango, then user gets 5+2 = 7 chances , as mango is a five letter word. 

## words to play on
### (apple  banana   mango   strawberry   orange   grape   pineapple   apricot   lemon   coconut   watermelon   cherry papaya   berry   peach   lychee   muskmelon)


```python
#to clean the input fruits we will play on
def clean_data(fruits):
    fruits = fruits.replace("\n"," ")
    list_of_fruits = fruits.split(" ")
    cleaned_fruits = []
    for i in range (len (list_of_fruits)):
        if len (list_of_fruits[i]) != 0:
            cleaned_fruits.append(list_of_fruits[i])
    return (cleaned_fruits)
            
```

```python
Enter_Fruits= input ("enter your fruits with space between every fruit: ")
fruits=clean_data(Enter_Fruits)
fruits
```
